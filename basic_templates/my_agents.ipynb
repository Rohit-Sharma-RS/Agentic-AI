```python
from dotenv import load_dotenv
from openai import OpenAI
import os
load_dotenv(override=True)

openrouter_api = os.getenv('OPEN_ROUTER_API')
groq_api = os.getenv('GROQ_API')
request = """Please come up with a challenging and nuanced question with a succint answer,
that I can ask a number of LLMs to evaluate there intelligence 
not a mathematical puzzle I want it to be thought provoking with intelligent insight.
Include in your question that the answer must be short"""

request+="Answer only with the question no explanation"

messages = [{'role': 'user', 'content': 'hi'}]
```

```python
openrouter_base_url = "https://openrouter.ai/api/v1"
groq_base_url = "https://api.groq.com/openai/v1"

openrouter = OpenAI(api_key=openrouter_api, base_url=openrouter_base_url)
groq = OpenAI(api_key=groq_api, base_url=groq_base_url)
```

```python
# Strong general-purpose choice
model_name = "deepseek/deepseek-chat-v3-0324"

response = openrouter.chat.completions.create(
    model=model_name,
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```

```python
model="openai/gpt-oss-20b"
response = groq.chat.completions.create(model=model, messages=messages)
print(response.choices[0].message.content)
```

```python
response = groq.chat.completions.create(
    model='openai/gpt-oss-20b',
    messages=messages
)

answer = response.choices[0].message.content

print(answer)

for i in ['autumn', 'diana', 'hannah', 'austin', 'daniel', 'troy']:
    print(i)
    tts = groq.audio.speech.create(
    model="canopylabs/orpheus-v1-english",
    voice=i,
    input=answer,
    response_format="wav"
)
    tts.write_to_file(f"response{i}.wav")

print("Audio saved as response.wav")
```
